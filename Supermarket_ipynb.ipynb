```python
s = df['Selling price']

q1 = s.quantile(0.25)
q3 = s.quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
print(f"Q1={q1:.1f},Q3={q3:.1f},IQR={iqr:.1f}")
print(f"Lower limit={lower:.1f},upper limit={upper:.1f}")

outliers= df[(s<lower)|(s>upper)]
print("outliers bills:",len(outliers))

outliers[['Date','Branch','Customer type','Gender','Product line',
          'Unit price','Quantity','Selling price','profit']] \
    .sort_values(by='Selling price', ascending=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,3.5))
sns.boxplot(x=df['Selling price'],color='#4C72B0')
plt.axvline(upper,color='green',linestyle='--',label=f'Upper limit{upper:.0f}')
plt.title('Bill value outliers')
plt.legend()
plt.show()
```
